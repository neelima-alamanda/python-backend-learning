## Task 4: Count Characters

Read a text file and display:

- Total number of characters
- Total number of lines

````python
# Initialize character and line counters
char_count = 0
line_count = 0

# Open the file in read mode
with open("notes.txt", "r") as file:
    # Read the file line by line
    for line in file:
        line_count += 1          # Count each line
        char_count += len(line)  # Count characters in each line

# Display the results
print("Total Characters:", char_count)
print("Total Lines:", line_count)
````

## Task 5: Count Words

Read a file and count the total number of words.

**Example Output**

```
Total Words: 45
```

````python
# Open the file in read mode
with open("notes.txt", "r") as file:
    # Read the entire file
    contents = file.read()

    # Split the contents into words
    words = contents.split()

    # Count the total number of words
    count = len(words)

# Display the total word count
print("Total Words:", count)
````

## Task 6: Read Line by Line

Read a file using a **for** loop and print each line along with its line number.

**Example Output**

```
Line 1: Python
Line 2: Java
Line 3: C++
```

````python
# Initialize the line number
line_number = 1

# Open the file in read mode
with open("notes.txt", "r") as file:
    # Read the file line by line
    for line in file:
        # Display the line number and its content
        print(f"Line {line_number}: {line.strip()}")

        # Increment the line number
        line_number += 1
````

## Task 7: Copy File Contents

Create a program that copies all the contents from **source.txt** into **destination.txt**.

````python
# Create and write data into the source file
with open("source.txt", "w") as file:
    file.write("This is the source file.\nNow copying into the destination file.")

# Open the source file in read mode
with open("source.txt", "r") as source:
    # Open the destination file in write mode
    with open("destination.txt", "w") as destination:
        # Copy the contents line by line
        for line in source:
            destination.write(line)

print("File copied successfully.")
````

## Task 9: Create a CSV File

Create a CSV file called **students.csv** with the following columns:

- Name
- Age
- Course

Add at least **five student records**.

````python
import csv

# Student data
data = [
    {'Name': 'Alice', 'Age': 19, 'Course': 'HTML'},
    {'Name': 'Bob', 'Age': 18, 'Course': 'CSS'},
    {'Name': 'Charlie', 'Age': 20, 'Course': 'JavaScript'},
    {'Name': 'Diana', 'Age': 21, 'Course': 'Python'},
    {'Name': 'Ethan', 'Age': 22, 'Course': 'Data Science'}
]

# Open the CSV file in write mode
with open("students.csv", "w", newline="") as file:
    # Define the column names
    fieldnames = ["Name", "Age", "Course"]

    # Create a DictWriter object
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    # Write the header and student records
    writer.writeheader()
    writer.writerows(data)

print("Student details added to students.csv successfully.")
````

## Task 14: Handle Multiple Exceptions

Write a program that accepts two numbers from the user.

Handle the following exceptions:

- ValueError
- ZeroDivisionError

````python
# Try to accept two numbers and perform division
try:
    n1 = int(input("Enter the first number: "))
    n2 = int(input("Enter the second number: "))

    division = n1 / n2
    print("Result:", division)

# Handle invalid input
except ValueError:
    print("Please enter valid numbers.")

# Handle division by zero
except ZeroDivisionError:
    print("Division cannot be performed by 0.")
````

## Task 15: Use else and finally

Write a program that divides two numbers.

Requirements:

- Handle exceptions
- Display the result using **else**
- Display

```
Program Ended
```

using **finally**.

````python
# Try to accept two numbers and perform division
try:
    n1 = int(input("Enter the first number: "))
    n2 = int(input("Enter the second number: "))

    division = n1 / n2

# Handle invalid input
except ValueError:
    print("Please enter valid numbers.")

# Handle division by zero
except ZeroDivisionError:
    print("Division cannot be performed by 0.")

# Display the result if no exception occurs
else:
    print("Result:", division)

# Execute this block whether an exception occurs or not
finally:
    print("Program Ended")
````

## Task 16: Custom Exception

Create a custom exception called **InvalidAgeError**.

Raise the exception if the user enters an age less than **18**.

**Example**

```
Enter Age: 15
```

Output:

```
InvalidAgeError: Age must be 18 or above.
```

````python
# Create a custom exception class
class InvalidAgeError(Exception):
    def __init__(self, message):
        super().__init__(message)

# Check the user's age
try:
    age = int(input("Enter your age: "))

    if age < 18:
        raise InvalidAgeError("Age must be 18 or above.")

    print("You are eligible.")

# Handle the custom exception
except InvalidAgeError as e:
    print("InvalidAgeError:", e)
````

## Task 20: Mini Student Record System ⭐

Create a **Student Record System** that performs the following operations:

- Store student details in a CSV file
- Add a new student
- Display all students
- Search a student by name
- Handle FileNotFoundError
- Handle invalid input using exception handling
- Use the **with** statement
- Use **csv.DictReader()** and **csv.DictWriter()**

````python
import csv

# File name
filename = "students.csv"

# Add a new student
try:
    name = input("Enter student name: ")
    age = int(input("Enter age: "))
    course = input("Enter course: ")

    with open(filename, "a", newline="") as file:
        fieldnames = ["Name", "Age", "Course"]
        writer = csv.DictWriter(file, fieldnames=fieldnames)

        # Write header if file is empty
        if file.tell() == 0:
            writer.writeheader()

        writer.writerow({
            "Name": name,
            "Age": age,
            "Course": course
        })

    print("Student added successfully.")

except ValueError:
    print("Invalid age. Please enter a number.")

# Display all students
try:
    print("\nStudent Records")

    with open(filename, "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            print(row["Name"], "-", row["Age"], "-", row["Course"])

except FileNotFoundError:
    print("File not found.")

# Search a student
try:
    search = input("\nEnter student name to search: ")

    with open(filename, "r") as file:
        reader = csv.DictReader(file)

        found = False

        for row in reader:
            if row["Name"].lower() == search.lower():
                print("Student Found")
                print(row)
                found = True
                break

        if not found:
            print("Student not found.")

except FileNotFoundError:
    print("File not found.")
````
