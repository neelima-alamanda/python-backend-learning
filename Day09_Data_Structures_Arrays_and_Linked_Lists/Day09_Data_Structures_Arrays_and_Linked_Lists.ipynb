## Task 9: Reverse an Array

Reverse the array **without using** `reverse()` or slicing.

**Input**

```python
[1, 2, 3, 4, 5]
```

**Expected Output**

```python
[5, 4, 3, 2, 1]
```

````python
# Create the array
array = [1, 2, 3, 4, 5]

# Create an empty array to store the reversed elements
reversed_array = []

# Reverse the array
for index in range(len(array) - 1, -1, -1):
    reversed_array.append(array[index])

# Display the reversed array
print("Reversed Array:", reversed_array)
````

## Task 10: Find Maximum and Minimum

Find both the maximum and minimum values using a **single traversal**.

**Input**

```python
[45, 78, 12, 99, 34]
```

**Expected Output**

```text
Maximum = 99
Minimum = 12
```

````python
# Create the array
array = [45, 78, 12, 99, 34]

# Initialize the minimum and maximum values
minimum = array[0]
maximum = array[0]

# Find the minimum and maximum values
for number in array:
    if number < minimum:
        minimum = number
    if number > maximum:
        maximum = number

# Display the results
print("Minimum =", minimum)
print("Maximum =", maximum)
````

## Task 18: Delete Middle Node

Delete **30** from the linked list.

**Input**

```text

```


Count the number of nodes in the linked list.

## Task 20: Find the Middle Node (Challenge ⭐)

Find the middle node of the linked list **without converting it into a Python list**.

**Input**

```text
10 → 20 → 30 → 40 → 50
```

**Expected Output**

```text
Middle Node = 30
```

**Hint:** Use the **Slow Pointer** and **Fast Pointer** technique.

````python
# Create the node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Create the linked list
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)
node5 = Node(50)

head = node1
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

# Initialize the slow and fast pointers
slow = head
fast = head

# Find the middle node
while fast and fast.next:
    slow = slow.next
    fast = fast.next.next

# Display the middle node
print("Middle Node =", slow.data)
````
