## Task 1: Student Class

Create a `Student` class with `name`, `roll_no`, and `marks`.

- `display_details()`
- `is_pass()` (Pass if marks >= 35)

```text

## Task 2: Bank Account

Create a `BankAccount` class with `account_number`, `holder_name`, and `balance`.

Methods:
- `deposit()`
- `withdraw()`
- `check_balance()`

**Note:** Do not allow withdrawal if the balance is insufficient.

````python
# Create the BankAccount class
class BankAccount:

    # Initialize bank account details
    def __init__(self, account_number, holder_name, balance=0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    # Deposit money into the account
    def deposit(self, amount):
        self.balance += amount

    # Withdraw money from the account
    def withdraw(self, amount):
        if self.balance < amount:
            return "Insufficient Balance"

        self.balance -= amount

    # Display the current balance
    def check_balance(self):
        return self.balance


# Create a BankAccount object
account = BankAccount(2456e7, "John", 1000)

print(account.withdraw(2000))

account.deposit(1000)
account.withdraw(2000)

print("Current Balance:", account.check_balance())
````

## Task 4: Library Book

Create a `LibraryBook` class with `book_name`, `author`, and `available=True`.

Methods:
- `borrow()`
- `return_book()`

**Note:** Display an error message if the book is not available for borrowing.

````python
# Create the LibraryBook class
class LibraryBook:

    # Initialize book details
    def __init__(self, book_name, author, available=True):
        self.book_name = book_name
        self.author = author
        self.available = available

    # Borrow the book
    def borrow(self):
        if self.available:
            self.available = False
            print(f"You borrowed '{self.book_name}'.")
        else:
            print(f"Sorry! '{self.book_name}' is not available.")

    # Return the book
    def return_book(self):
        if not self.available:
            self.available = True
            print(f"You returned '{self.book_name}'.")
        else:
            print(f"'{self.book_name}' was not borrowed.")


# Create a LibraryBook object
book = LibraryBook("Harry Potter", "J.K. Rowling")

book.borrow()
book.borrow()

book.return_book()
book.return_book()
````

## Task 5: Animal Hierarchy

Create an `Animal` class with a `sound()` method.

Create the following subclasses:
- `Dog`
- `Cat`
- `Cow`

Override the `sound()` method in each subclass and use a loop to demonstrate polymorphism.

````python
# Create the Animal class
class Animal:

    # Display the animal sound
    def sound(self):
        print("Animal sound")


# Create the Dog class
class Dog(Animal):

    # Override the sound method
    def sound(self):
        print("Dog barks")


# Create the Cat class
class Cat(Animal):

    # Override the sound method
    def sound(self):
        print("Cat meows")


# Create the Cow class
class Cow(Animal):

    # Override the sound method
    def sound(self):
        print("Cow moos")


# Create a list of animal objects
animals = [Dog(), Cat(), Cow()]

# Demonstrate polymorphism using a loop
for animal in animals:
    animal.sound()
````

## Task 8: Shape Calculator

Create a `Shape` class with an `area()` method.

Create the following subclasses:
- `Rectangle`
- `Circle`
- `Triangle`

Override the `area()` method in each subclass and calculate the area using polymorphism.

````python
# Import the math module
import math

# Create the Shape class
class Shape:

    # Define the area method
    def area(self):
        pass


# Create the Rectangle class
class Rectangle(Shape):

    # Initialize rectangle dimensions
    def __init__(self, length, width):
        self.length = length
        self.width = width

    # Calculate rectangle area
    def area(self):
        return self.length * self.width


# Create the Circle class
class Circle(Shape):

    # Initialize circle radius
    def __init__(self, radius):
        self.radius = radius

    # Calculate circle area
    def area(self):
        return math.pi * self.radius ** 2


# Create the Triangle class
class Triangle(Shape):

    # Initialize triangle dimensions
    def __init__(self, base, height):
        self.base = base
        self.height = height

    # Calculate triangle area
    def area(self):
        return 0.5 * self.base * self.height


# Create a list of shape objects
shapes = [
    Rectangle(10, 5),
    Circle(7),
    Triangle(6, 4)
]

# Display the area of each shape
for shape in shapes:
    print(f"{shape.__class__.__name__} Area: {shape.area()}")
````

## Task 9: ATM

Create an `ATM` class with a private attribute `__balance`.

Methods:
- `deposit()`
- `withdraw()`
- `get_balance()`

**Note:** Do not access the private attribute directly.

````python
# Create the ATM class
class ATM:

    # Initialize the private balance
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance

    # Deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Invalid deposit amount")

    # Withdraw money
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance")

    # Return the current balance
    def get_balance(self):
        return self.__balance


# Create an ATM object
atm = ATM(1000)

# Perform transactions
atm.deposit(500)
atm.withdraw(2000)
atm.withdraw(1200)

# Display the current balance
print("Current Balance:", atm.get_balance())
````

## Task 11: Class Method

Create an `Employee` class.

Use a class method `from_string()` to create an object from the string:

`"Rahul-45000"`

Display the employee details.

````python
# Create the Employee class
class Employee:

    # Initialize employee details
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # Create an object from a string
    @classmethod
    def from_string(cls, emp_str):
        name, salary = emp_str.split("-")
        return cls(name, int(salary))


# Create an Employee object using the class method
employee = Employee.from_string("Rahul-45000")

# Display employee details
print("Name   :", employee.name)
print("Salary :", employee.salary)
````

## Task 12: Static Method

Create a `Calculator` class.

Create the following static methods:
- `is_even()`
- `is_prime()`
- `factorial()`

Call all the methods directly using the class name without creating an object.

````python
# Create the Calculator class
class Calculator:

    # Check whether a number is even
    @staticmethod
    def is_even(num):
        if num % 2 == 0:
            return True
        return False

    # Check whether a number is prime
    @staticmethod
    def is_prime(num):
        if num < 2:
            return False

        for i in range(2, num):
            if num % i == 0:
                return False
        return True

    # Calculate the factorial of a number
    @staticmethod
    def factorial(num):
        fact = 1
        for i in range(1, num + 1):
            fact *= i
        return fact


# Call static methods using the class name
print("Even Checker      :", Calculator.is_even(89))
print("Prime Checker     :", Calculator.is_prime(89))
print("Factorial of 5    :", Calculator.factorial(5))
````

## Task 13: Logging Decorator

Create a decorator that prints:

- `Starting Function...`
- `Ending Function...`

before and after executing any function.

Apply the decorator to:
- `greet()`
- `add()`
- `login()`

````python
# Create a logging decorator
def log_decorator(func):

    # Wrapper function
    def wrapper(*args, **kwargs):
        print("\nStarting Function...")
        result = func(*args, **kwargs)
        print("Ending Function...")
        return result

    return wrapper


# Decorate the greet function
@log_decorator
def greet():
    print("Hello!")


# Decorate the add function
@log_decorator
def add(a, b):
    print(a + b)


# Decorate the login function
@log_decorator
def login():
    print("Login Successful")


# Call the functions
greet()
add(2, 3)
login()
````

## Task 14: Timer Decorator

Create a decorator that displays:

- `Execution Time:`

Use the `time.time()` function to calculate the execution time.

Apply the decorator to:
- `calculate_sum()`

````python
import time

# Create a timer decorator
def timer_decorator(func):

    # Wrapper function
    def wrapper(*args, **kwargs):
        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()
        print("Execution Time:", end - start, "seconds")

        return result

    return wrapper


# Decorate the calculate_sum function
@timer_decorator
def calculate_sum():
    total = 0

    for i in range(1, 1000000):
        total += i

    print("Sum is:", total)


# Call the function
calculate_sum()
````

## Task 15: Authentication Decorator

Create a variable:

- `is_logged_in = True`

Create an authentication decorator that:

- Executes the function if the user is logged in.
- Otherwise, prints:
  - `Access Denied`

Apply the decorator to:
- `view_profile()`
- `delete_account()`
```

````python
# Login status
is_logged_in = True

# Create an authentication decorator
def authentication_decorator(func):

    # Wrapper function
    def wrapper(*args, **kwargs):
        if is_logged_in:
            return func(*args, **kwargs)
        else:
            print("Access Denied")

    return wrapper


# Decorate the view_profile function
@authentication_decorator
def view_profile():
    print("Profile Details...")


# Decorate the delete_account function
@authentication_decorator
def delete_account():
    print("Account Deleted")


# Call the functions
view_profile()
delete_account()
````

## Task 17: Fibonacci Generator

Create a generator that yields the Fibonacci sequence:

- `0`
- `1`
- `1`
- `2`
- `3`
- `5`
- `8`
- ...

Generate and display the **first 15 Fibonacci numbers** using the generator.

````python
# Create a Fibonacci generator
def fib_gen(n):
    a, b = 0, 1

    for _ in range(n):
        yield a
        a, b = b, a + b


# Create the generator object
generator = fib_gen(15)

# Display the first 15 Fibonacci numbers
for number in generator:
    print(number)
````

## Task 18: Custom Context Manager

Create your own context manager named `Logger`.

- When entering the context (`__enter__`), print:
  - `Opening Resource...`

- When exiting the context (`__exit__`), print:
  - `Closing Resource...`

Inside the `with` block, print:
- `Working...`

````python
# Create a custom context manager
class Logger:

    # Runs when entering the with block
    def __enter__(self):
        print("Opening Resource...")
        return self

    # Runs when exiting the with block
    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing Resource...")


# Use the context manager
with Logger():
    print("Working...")
````

## Task 19: Regex Validation

Given the following text:

```python
text = """
rahul@gmail.com
abc@yahoo.com
12345
9876543210
https://google.com
"""
```

Using Regular Expressions (`re` module), find:

- Email IDs
- Phone Numbers
- URLs

````python
import re

# Sample text
text = """
rahul@gmail.com
abc@yahoo.com
12345
9876543210
https://google.com
"""

# Find email IDs
emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

# Find 10-digit phone numbers
phone_numbers = re.findall(r'\d{10}', text)

# Find URLs
urls = re.findall(r'https?://\S+', text)

# Display the results
print("Email IDs:", emails)
print("Phone Numbers:", phone_numbers)
print("URLs:", urls)
````

## Task 20: Mini Utility Program

Write one Python program that uses **at least six standard libraries**.

### Requirements

- Create a folder using `os`.
- Store student data in JSON format using `json`.
- Generate a random student ID using `random`.
- Store today's date using `datetime`.
- Calculate GPA percentage using `math`.
- Count repeated grades using `collections.Counter`.
- Save all the student information into an SQLite database using `sqlite3`.

Finally, display:

```text
Student Added Successfully
```

````python
import os
import json
import random
from datetime import datetime
import math
from collections import Counter
import sqlite3

# Create a folder
folder_name = "Student_Data"
os.makedirs(folder_name, exist_ok=True)

# Get student details
name = input("Enter Student Name: ")
gpa = float(input("Enter GPA: "))
grades = input("Enter Grades (separated by spaces): ").split()

# Generate a random student ID
student_id = random.randint(1000, 9999)

# Get today's date
today = datetime.now().strftime("%d-%m-%Y")

# Calculate percentage
percentage = math.ceil(gpa * 9.5)

# Count repeated grades
grade_count = Counter(grades)

# Store student data
student = {
    "Student ID": student_id,
    "Name": name,
    "Date": today,
    "GPA": gpa,
    "Percentage": percentage,
    "Grades": grade_count
}

# Save data to a JSON file
with open(f"{folder_name}/students.json", "w") as file:
    json.dump(student, file, indent=4)

# Connect to SQLite database
conn = sqlite3.connect(f"{folder_name}/students.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
    student_id INTEGER,
    name TEXT,
    date TEXT,
    gpa REAL,
    percentage REAL
)
""")

# Insert student data
cursor.execute(
    "INSERT INTO students VALUES (?, ?, ?, ?, ?)",
    (student_id, name, today, gpa, percentage)
)

# Save changes and close the database
conn.commit()
conn.close()

# Display success message
print("Student Added Successfully")
````
