## Task 11: Implement Stack

Create a `Stack` class with the following methods:

- `push()`
- `pop()`
- `peek()`
- `is_empty()`
- `display()`

````python
# Create the Stack class
class Stack:
    def __init__(self):
        self.stack = []

    # Add an element to the stack
    def push(self, data):
        self.stack.append(data)

    # Remove the top element
    def pop(self):
        return self.stack.pop()

    # Check if the stack is empty
    def is_empty(self):
        return self.stack == []

    # Display the stack
    def display(self):
        print(self.stack)

    # View the top element
    def peek(self):
        return self.stack[-1]


# Create a Stack object
stack = Stack()

# Push elements into the stack
stack.push(10)
stack.push(20)
stack.push(30)

# Display the stack
stack.display()

# View and remove the top element
print("Top Element:", stack.peek())
print("Popped Element:", stack.pop())

# Display the updated stack
stack.display()

# Check if the stack is empty
print("Is Empty:", stack.is_empty())
````

## Task 12: Implement Queue

Create a `Queue` class using a **list**.

Implement the following methods:

- `enqueue()`
- `dequeue()`
- `peek()`
- `is_empty()`
- `display()`

````python
# Create the Queue class
class Queue:
    def __init__(self):
        self.queue = []

    # Add an element to the queue
    def enqueue(self, data):
        self.queue.append(data)

    # Remove the front element
    def dequeue(self):
        if self.is_empty():
            return "Queue is empty"
        return self.queue.pop(0)

    # View the front element
    def peek(self):
        return self.queue[0]

    # Check if the queue is empty
    def is_empty(self):
        return self.queue == []

    # Display the queue
    def display(self):
        print(self.queue)


# Create a Queue object
queue = Queue()

# Try to dequeue from an empty queue
print(queue.dequeue())

# Add elements to the queue
queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)

# Display the queue
queue.display()

# View and remove the front element
print("Peek Element:", queue.peek())
print("Dequeued Element:", queue.dequeue())

# Display the updated queue
queue.display()

# Check if the queue is empty
print("Is Empty:", queue.is_empty())
````

## Task 17: Balanced Parentheses

Write a program to check whether an expression has **balanced parentheses**.

**Examples**

```text
((())) → True
(()() → False
```

**Requirement:** Use a **Stack**.

````python
# Get the expression from the user
expression = input("Enter expression to check balanced: ")

# Create an empty stack
stack = []
balanced = True

# Check each character in the expression
for character in expression:
    if character == '{' or character == '[' or character == '(':
        stack.append(character)
    else:
        # Check if there is no opening bracket
        if stack == []:
            balanced = False
            break

        # Check whether the closing bracket matches the top of the stack
        top = stack.pop()

        if character == '}' and top != '{':
            balanced = False
            break
        elif character == ']' and top != '[':
            balanced = False
            break
        elif character == ')' and top != '(':
            balanced = False
            break

# Check if any opening brackets are still left
if stack:
    balanced = False

# Display the result
if balanced:
    print("Expression is balanced")
else:
    print("Expression is not balanced")
````

## Task 18: Printer Queue

Create a **Queue** to manage print jobs.

Display the following menu:

```text
1. Add Print Job
2. Print Document
3. View Next Document
4. Exit
```

Implement the printer queue using a **Queue**.

````python
# Create an empty queue
queue = []

# Display the printer queue menu
while True:
    print("\n1. Add Print Job")
    print("2. Print Document")
    print("3. View Next Document")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    match choice:
        case 1:
            # Add a print job to the queue
            job = input("Enter Job that you want to print: ")
            queue.append(job)

        case 2:
            # Print and remove the first document
            if queue == []:
                print("Queue is empty")
            else:
                print("Printing:", queue.pop(0))

        case 3:
            # View the next document without removing it
            if queue == []:
                print("Queue is empty")
            else:
                print("Next Document:", queue[0])

        case 4:
            # Exit the program
            print("Exiting...")
            break

        case _:
            # Handle invalid menu choices
            print("Invalid choice")
````

## Task 19: Hospital Management System

Each patient has:

- **Name**
- **Priority**

### Rules

- Higher priority patients are treated first.
- If the priority is the same, patients are treated in **FIFO** order.
- Display the treatment order.

### Example Input

```text
Ram 2
Sita 5
Hari 3
Anu 5
```

### Expected Output

```text
Sita
Anu
Hari
Ram
```

````python
# Create a dictionary to store patient names and priorities
queue = {}

# Add patients to the queue
queue["Ram"] = 2
queue["Sita"] = 5
queue["Hari"] = 3
queue["Anu"] = 5

# Display patients in treatment order
while queue:
    # Find the highest priority
    highest_priority = max(queue.values())

    # Find the first patient with the highest priority
    for name, priority in queue.items():
        if priority == highest_priority:
            print(name)
            queue.pop(name)
            break
````

## Task 20: Circular Queue Menu Program

Implement a **Circular Queue** with a menu.

```text
1. Enqueue
2. Dequeue
3. Peek
4. Display
5. Exit
```

**Queue Size:** `5`

Requirements:

- Display **"Queue Full"** when the queue is full.
- Display **"Queue Empty"** when the queue is empty.

````python
# Set the queue size
size = 5

# Create an empty circular queue
queue = [None] * size

# Initialize front, rear, and count
front = 0
rear = 0
count = 0

# Display the menu repeatedly
while True:
    print("\n1. Enqueue")
    print("2. Dequeue")
    print("3. Peek")
    print("4. Display")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    match choice:
        case 1:
            # Add an element to the circular queue
            if count == size:
                print("Queue Full")
            else:
                data = int(input("Enter the data: "))
                queue[rear] = data
                rear = (rear + 1) % size
                count += 1

        case 2:
            # Remove the front element
            if count == 0:
                print("Queue Empty")
            else:
                print("Dequeued Element:", queue[front])
                queue[front] = None
                front = (front + 1) % size
                count -= 1

        case 3:
            # View the front element
            if count == 0:
                print("Queue Empty")
            else:
                print("Front Element:", queue[front])

        case 4:
            # Display the circular queue
            if count == 0:
                print("Queue Empty")
                print(queue)
            else:
                print(f"Queue: {queue}")
                print(f"Front: {front}, Rear: {rear}")

        case 5:
            # Exit the program
            print("Exiting...")
            break

        case _:
            # Handle invalid choices
            print("Invalid choice")
````
